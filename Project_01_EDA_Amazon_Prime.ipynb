# **8. Data Wrangling**

**8.1 Column Handling**

```python
# Drop show_id column
data = data.drop(columns=['show_id'])
```

```python
# Rename listed_in column to genre
data = data.rename(columns={'listed_in': 'genre'})
```

**8.2 Feature Creation**

```python
# duration_value – Extracted numerical values from the duration column for better comparison and analysis.
duration_values = []

for d in data['duration']:
    if pd.notnull(d):
        duration_values.append(int(d.split()[0]))
    else:
        duration_values.append(None)

data['duration_value'] = duration_values
```

```python
data.columns
```

**8.3 Date & Time Feature Extraction**

```python
# Convert to datetime format
data['date_added'] = pd.to_datetime(data['date_added'], errors='coerce')
```

```python
# Extract Year – To analyze yearly trends in content addition
data['year_added'] = data['date_added'].dt.year
```

```python
# Extract Month – To study seasonal patterns in content uploads
data['month_added'] = data['date_added'].dt.month
```

```python
# Extract Day – To observe daily distribution
data['day_added'] = data['date_added'].dt.day
```

```python
# Extract Week – To identify weekly content addition behavior
data['week_added'] = data['date_added'].dt.isocalendar().week
```

```python
# Extract Quarter – To analyze quarterly trends in platform expansion
data['quarter_added'] = data['date_added'].dt.quarter
```

```python
data.columns
```

**8.4 Data Type Conversion**

```python
# type, genre → Converted to categorical to improve analysis clarity.
cat_cols = ['type', 'genre']
data[cat_cols] = data[cat_cols].astype('category')
```

**Insights**

**Insights**
